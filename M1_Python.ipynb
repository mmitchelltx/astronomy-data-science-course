## 8) Data Structure - List

```python
# use [] to store elements in a list
names_student = ['Josh', 'Zach', 'Jenny','Marsha', 'Amber', "Gary"]
```

```python
print(names_student)
```

```python
print(type(names_student))
```

```python
# you can store any data type or structure in a list
list_example = ['this is a string', 12345, 1.2, [1.5, 'this is another string to test a list in a list']]
```

```python
print(list_example)
print(type(list_example))
```

```python
# with indexing i can access any element from the list
# indexing from left starts at 0, from the right starts at -1

print(list_example[-1])
```

```python
# extract 2 from string2 using indexing, detail extends down to the character
list_example
print(list_example[-1][-1][-6])
```

```python
# extract "s" from is from example
print(list_example[-1][-1][6])
```

```python
# slicing exersize, exact ny from Jenny
print(names_student[2])
```

```python
print(names_student[2][3:5]) # extend past the desired charater to include
```

```python
list_ex = [1,2,3,4,5]

# extract 3,4,5 from list
print(list_ex[2::1]) #if you leave ending index empty it goes to last element possible


# extract 1,3
print(list_ex[:3:2]) # if starting index empty it starts from first possible character

# reverse entire list
print(list_ex[::-1])
```

```python
# lists are mutable - you can change any element from the list at anytime

# replace Zach with Harold
names_student[1] = 'Harold'
```
