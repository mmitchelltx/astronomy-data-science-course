## 2) Data Type - String

```python
# Use single quotes to create a string
'this is a string, 513'
'this is also a string'
```

```python
# use print command to display both strings
print('this is a string, 513')
print('this is also a string')
```

```python
# type command displays the type of data
type('is this a string')
```

```python
# you can also combine commands
print(type('is this a string'))
```

```python
# double quotes can also give you a string
"This is also a string!"
```

```python
len('this is a string, len gives us a count of the charaters')
```

## 3) Data Type - Variables and Constants

```python
# to store data that changes with time, use variables
age = 35 # variables in python are always written with lowercase letters
```

```python
# display type of data in variable
type(35)
```

```python
# to store something that can never change with time use a constant
# create constants using UPPER CASE letters
PI = 3.14 # pi value
G = 9.81 # constant for gravity
```

## 4) Data Type - Numerical
* Integers
* Floats
* Complex Numbers

```python
# fntegers is a whole number, can be negative or positive
print(type(7))
print(type(49))
print(type(-7))
```

```python
# floats are can be any number with a .
print(type(7.0))
print(type(-87.938473))
print(type(97.838292))
```

```python
# complex numbers will have an inclusion of j for imaginary part of complex number
print(type(3j))
print(type(3j+4))

```
